# Data Preparation
Validate road images and their damage annotations.


```python
from pathlib import Path
import sys

# Resolve shared modules and portable project settings.
ROOT=next((p for p in [Path.cwd(),*Path.cwd().parents] if (p/'configs/project.json').is_file()),None)
if ROOT is None:
    raise FileNotFoundError('RoadScan project configuration was not found.')
if str(ROOT/'src') not in sys.path:
    sys.path.insert(0,str(ROOT/'src'))
from roadscan.settings import load_settings
settings=load_settings(ROOT)
```

```python
import json
import pandas as pd
from roadscan.preprocessing import validate_dataset

# Verify paired files, class mappings, coordinates and recorded checksums.
result=validate_dataset(settings['dataset_root'])
out=ROOT/'results/data_validation';out.mkdir(parents=True,exist_ok=True)
pd.DataFrame(result['summary']).to_csv(out/'split_summary.csv',index=False)
pd.DataFrame(result['classes']).to_csv(out/'class_distribution.csv',index=False)
(out/'validation.json').write_text(json.dumps(result,indent=2),encoding='utf-8')
display(pd.DataFrame(result['summary']))
```
