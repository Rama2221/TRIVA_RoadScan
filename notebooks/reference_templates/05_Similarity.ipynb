# Report Similarity
Represent detected damage with frozen visual features for candidate retrieval.


```python
from pathlib import Path
import sys

# Resolve shared modules and portable project settings.
ROOT=next((p for p in [Path.cwd(),*Path.cwd().parents] if (p/'configs/project.json').is_file()),None)
if ROOT is None:
    raise FileNotFoundError('RoadScan project configuration was not found.')
if str(ROOT/'src') not in sys.path:
    sys.path.insert(0,str(ROOT/'src'))
from roadscan.settings import load_settings
settings=load_settings(ROOT)
```

```python
import json
from roadscan.service import DINOEncoder

# Keep the retrieval model frozen and reuse its image processor.
encoder_path=Path(settings['deployment_package'])/'dinov2_encoder'
encoder=DINOEncoder(encoder_path,device=settings['device'])
summary=dict(model='DINOv2 Small',embedding_dimension=encoder.dimension,
    representation='L2-normalized CLS features',similarity='cosine',
    evaluation='Distinct-view labelled duplicate pairs are not available.')
out=ROOT/'results/similarity';out.mkdir(parents=True,exist_ok=True)
(out/'representation.json').write_text(json.dumps(summary,indent=2),encoding='utf-8')
display(summary)
```
